# Resolvendo a atividade 5 em células

```python
import pandas as pd
import numpy as np

mercado_df = pd.read_csv("vendas_mercados.csv")
print(mercado_df)
```

### 1. Considerando que o Lucro é calculado pela diferença entre o Preço Unitário e o Custo Unitário, multiplicada pela Quantidade vendida: Qual mercado obteve o maior lucro total ao final do período?

```python
mercado_df["lucro"] = ((mercado_df["Preco_Unitario"] - mercado_df["Custo_Unitario"]) * mercado_df["Quantidade"])

lucro_por_mercado = mercado_df.groupby(["Mercado"])["lucro"].sum()

print(lucro_por_mercado)

```

O mercado com maior lucro foi o Mercado Central com lucro de R$583,41

### 2. Qual forma de pagamento é mais utilizada pelos clientes em cada uma das regiões (Centro, Sul, Norte, Leste, Oeste)?

```python
pagamento_por_regiao = mercado_df.groupby(["Regiao", "Forma_Pagamento"])["ID_Venda"].nunique().reset_index()

mais_utilizada = (
    pagamento_por_regiao.sort_values(
        by=["Regiao", "ID_Venda"], ascending=[True, False]
    )
    .groupby("Regiao")
    .first()
    .sort_values(by="ID_Venda", ascending=False)
)

print(mais_utilizada)
```

### 3. De todos os mercados somados, qual foi o produto mais vendido (em quantidade de itens) e o produto menos vendido?

```python
produtos_vendidos = mercado_df.groupby("Produto")["Quantidade"].sum().sort_values(ascending=False)
print(produtos_vendidos)
mais_vendido = produtos_vendidos.head(1)
menos_vendido = produtos_vendidos.tail(1)
print("===== Mais vendido =====")
print(mais_vendido)

print("===== Menos vendido =====")
print(menos_vendido)

```
