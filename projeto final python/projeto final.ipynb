```python
import pandas as pd
import matplotlib as plt
# Configurações para melhorar a visualização das tabelas no notebook
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)

# 1. Carregamento da base respeitando o separador ';' e encoding 'utf-8'
df = pd.read_csv('vendas.csv', sep=';', encoding='utf-8')
```

```python
# 2. Dimensões da base (linhas e colunas)
print(
    f'Total de registros (linhas): {df.shape[0]} | Total de colunas:'f'{df.shape[1]}\n'
)

# 3. Visualizando as primeiras 5 linhas para entender a estrutura
display(df.head())

```

```python
# 4. Tipos de dados e contagem de valores não-nulos
print('\n--- Informações Estruturais e Tipos de Dados ---')
df.info()

# 5. Verificação inicial de qualidade: valores ausentes e duplicados
print('\n--- Contagem de Valores Ausentes (Nulos) ---')
print(df.isnull().sum())

print(f'\nTotal de linhas exatamente duplicadas: {df.duplicated().sum()}')
```

```python
# 1. Verificar linhas duplicadas (completas e por id_item_venda)
print(f'Linhas totalmente duplicadas: {df.duplicated().sum()}')
print(
    'IDs de item de venda duplicados:'
    f" {df.duplicated(subset=['id_item_venda']).sum()}"
)

# 2. Verificar valores únicos das colunas categóricas (procurando erros de digitação/padronização)
colunas_categoricas = [
    'categoria_produto',
    'estado_cliente',
    'nome_vendedor',
    'forma_pagamento',
    'status_venda',
]

for col in colunas_categoricas:
    print(f'\n--- Valores únicos em {col} ---')
    print(df[col].value_counts(dropna=False))

# 3. Verificar estatísticas das colunas numéricas (procurando zeros, negativos ou valores absurdos)
print('\n--- Resumo Estatístico das Colunas Numéricas ---')
display(df[['quantidade', 'preco_unitario', 'desconto', 'valor_total']].describe())

# 4. Testar conversão de data para ver se existem datas inválidas
datas_teste = pd.to_datetime(df['data_venda'], errors='coerce')
print(
    '\nDatas inválidas encontradas na conversão:'
    f' {datas_teste.isnull().sum()}'
)
if datas_teste.isnull().sum() > 0:
    print('Exemplos de datas inválidas')
    print(df.loc[datas_teste.isnull(), 'data_venda'].unique()[:10])
```
